```python
import sqlite3

# Create/connect DB
conn = sqlite3.connect('demo.db')
c = conn.cursor()

# Create table
c.execute('''CREATE TABLE IF NOT EXISTS households
             (id INTEGER PRIMARY KEY, gapaname TEXT, owned INTEGER, total INTEGER)''')

# Insert data
c.execute("INSERT OR IGNORE INTO households VALUES (1, 'Demo Gaunpalika', 100, 150)")

# Query
c.execute("SELECT * FROM households")
print(c.fetchall())  # Output: [(1, 'Demo Gaunpalika', 100, 150)]

conn.commit()
conn.close()
```

```python
# Pandas way – handles nesting automatically
import requests
import pandas as pd

# Fetch
r = requests.get('https://pokeapi.co/api/v2/pokemon/pikachu')
data = r.json()

# Process
# Manual extraction – understand the JSON
types = [t['type']['name'] for t in data['types']]
df = pd.DataFrame([{
    'name': data['name'],
    'height': data['height'],
    'weight': data['weight'],
    'types': types
}])
df
# Output: DataFrame with Pikachu stats

df_full = pd.json_normalize(data)
print(df_full.columns[:20])  # Shows all flattened columns
df_full[['name', 'height', 'weight', 'types']].head()
```

```python
selected = {
    'name': data['name'],
    'height': data['height'],
    'weight': data['weight'],
    'base_experience': data['base_experience'],
    'types': ', '.join(t['type']['name'] for t in data['types'])
}
df_clean = pd.DataFrame([selected])
df_clean
```

Exercise 1 – Simple Calculator
Write code that:
- Asks the user to input two numbers (use input())
- Asks which operation: +, −, ×, ÷
- Prints the result with 2 decimal places
- If user types anything else → print “Invalid operation”

```
x=int(input("enter the 1st number"))
y=int(input("enter the 2nd number"))
def add(a,b):
  return a+b

def sub(a,b):
  return a-b

def mul(a,b):
  return a*b

def div(a,b):
  return a/b

ch=input("enter the ur choice:a,b,c,d,e")
if ch == "a":
  print(f"{add(x,y):.2f}")
elif ch == "b":
  print(sub(x,y))
elif ch == "c":
  print(mul(x,y))
elif ch == "d":
  print(div(x,y))
else:
  print("invalid operation")
```

Exercise 2 – Grade Processor (Medium )
Create a program that:
1. Has a list marks = [78, 85, 92, 66, 88, 95, 71, 83]
2. Calculates and prints:
- Average mark
- Highest and lowest mark
- How many students got ≥ 80 (distinction)
3. Creates a new list grades where:
≥ 90 → “A+”, ≥ 80 → “A”, ≥ 70 → “B”, else → “C”
4. Saves the output in a file grade_report.txt exactly like this:

```
#1
marks = [78, 85, 92, 66, 88, 95, 71, 83]
#2
average = sum(marks) / len(marks)
highest = max(marks)
lowest = min(marks)
dis=0
for i in marks:
  if i>=80:
    dis+=1
print(f"Average mark: {average}")
print(f"Highest mark: {highest}")
print(f"Lowest mark: {lowest}")
print(f"student with distintion: {dis}")
```

```
#3
new= []
for i in marks:
  if i>=90:
    new.append("A+")
  elif i>=80:
    new.append("A")
  elif i>=70:
    new.append("B")
  else:
    new.append("C")
print(new)
```

```
#4
with open("marks.txt","w") as f:
  f.write(f"Average mark: {average}\n")
with open("marks.txt","a") as f:
  f.write(f"Highest mark: {highest}\n")
with open("marks.txt","a") as f:
  f.write(f"Lowest mark: {lowest}\n")
with open("marks.txt","a") as f:
  f.write(f"student with distintion: {dis}\n")
with open("marks.txt","a") as f:
  f.write(f"{new}")
```
